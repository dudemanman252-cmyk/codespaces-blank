```python
# Imports
from datetime import datetime
```

```python
# Input / Output
my_first_name = input("Enter your first name: ")
my_last_name = input("Enter your last name: ")
my_full_name = f"{my_first_name} {my_last_name}"
print(f"My full name is {my_full_name}")
```

```python
# Quotes
print('The backslash character "\\" is special. Don\'t forget to escape it!')
print('''"Programming isn't about what you know;
it's about what you can figure out."
- Chris Pine''')
```

```python
# Data Types
my_string = "5"
print(f"my_string: {type(my_string)}, {my_string}")
my_int = int(my_string)
print(f"my_int: {type(my_int)}, {my_int}")
my_float = float(my_int)
print(f"my_float: {type(my_float)}, {my_float}")
```

```python
# Boolean
my_false_var = 0
print(f"my_false_var: {type(my_false_var)}, {my_false_var}")
my_false_var = bool(my_false_var)
print(f"my_false_var: {type(my_false_var)}, {my_false_var}")
my_true_var = 1
print(f"my_true_var: {type(my_true_var)}, {my_true_var}")
my_true_var = bool(my_true_var)
print(f"my_true_var: {type(my_true_var)}, {my_true_var}")
```

```python
# Datetime
my_datetime_string = "05/03/2025"
print(f"my_datetime_string: {type(my_datetime_string)}, {my_datetime_string}")
my_datetime = datetime.strptime(my_datetime_string, "%m/%d/%Y")
print(f"my_datetime: {type(my_datetime)}, {my_datetime}")
my_datetime_string = my_datetime.strftime("%m-%d-%Y")
print(f"my_datetime_string: {type(my_datetime_string)}, {my_datetime_string}")
```
