## Comprehension\nCreate a list of squared even numbers without hard-coding the results.

```python
numbers = range(1, 11)\nsquared_even_numbers = [number ** 2 for number in numbers if number % 2 == 0]\nprint(squared_even_numbers)
```

## Lambda\nUse a lambda function to sort shapes by area.

```python
shapes = [Shape(3, 4), Shape(5, 2), Shape(6, 6)]\nsorted_shapes = sorted(shapes, key=lambda item: item.area())\nfor item in sorted_shapes:\n    print(item, item.area())
```

## Recursion\nUse recursion to calculate a factorial.

```python
def factorial(number):\n    if number <= 1:\n        return 1\n    return number * factorial(number - 1)\n\nvalue = 5\nprint(f"{value}! = {factorial(value)}")
```
